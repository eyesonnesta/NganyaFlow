# **NganyaFlow Dataset Exploration**

Selected Dataset: MTA M15 Hourly Ridership 2025 Q1

Expected output variable: ridership

The ridership variable represents passenger demand and can serve as the target variable for NganyaFlow's demand forecasting component.

```python
import pandas as pd
from pathlib import Path

ROOT = Path.cwd().resolve()
if not (ROOT / 'data/raw').is_dir():
    ROOT = ROOT.parent
assert (ROOT / 'data/raw').is_dir(), 'Open this notebook from its repository or notebooks folder.'
df = pd.read_csv(ROOT / 'data/raw/mta_ridership/MTA_M15_Hourly_Ridership_2025_Q1.csv')

df.head()
```

a. How many rows and columns are contained in the dataset?

```python
rows, columns = df.shape
print("No. of rows: ", rows)
print("No. of columns: ", columns)
```

The dataset contains 51,840 rows and 6 columns.

b. What datatypes are contained in the dataset?

```python
print(df.dtypes)
```

The dataset contains object and integer datatypes.

c. Is the dataset complete?

```python
df.isnull().sum()

total_missing = df.isnull().sum().sum()
print("Total Missing Values: ", total_missing)
if total_missing == 0:
  print("The dataset is complete. Yay!")
else:
  print("The dataset contains missing values.")
```

The dataset is complete because there are no missing values in any of its columns.

d. Slice out the first 15 rows and last 20 rows and merge them into a new dataframe titled "df_sample".

```python
first_15 = df.head(15)
last_20 = df.tail(20)

df_sample = pd.concat([first_15, last_20], ignore_index= True)

df_sample
```
